```python
# Task 1: Import required Python libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
from mlxtend.frequent_patterns import apriori, association_rules
print("Rohith John 24BAD100")
# Task 2: Load transactional dataset
# Replace 'groceries_dataset.csv' with the actual path to your downloaded file
file_path = 'Groceries_dataset.csv'
df = pd.read_csv(file_path)

print("Dataset Preview:")
print(df.head(), "\n")

# Task 3: Perform data preprocessing (convert to one-hot encoding)
# For the Kaggle dataset, a transaction is a combination of Member_number and Date.
basket = (df.groupby(['Member_number', 'Date', 'itemDescription'])['itemDescription']
          .count().unstack().reset_index().fillna(0)
          .set_index(['Member_number', 'Date']))

# Convert quantities to boolean values
def encode_units(x):
    if x <= 0:
        return False
    if x >= 1:
        return True

basket_encoded = basket.map(encode_units)
print(f"Encoded dataset shape (Transactions, Unique Items): {basket_encoded.shape}\n")

# Task 4 & 5: Generate frequent itemsets & set minimum support threshold
# Using 0.005 as minimum support for this specific retail dataset
min_support_threshold = 0.005
frequent_itemsets = apriori(basket_encoded, min_support=min_support_threshold, use_colnames=True)

# Add a length column for analysis
frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(lambda x: len(x))

print("Top 5 Frequent Itemsets by Support:")
print(frequent_itemsets.sort_values('support', ascending=False).head(), "\n")

# Task 6 & 7: Generate association rules and filter using confidence/lift
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

# Filter for confidence >= 0.1 and lift > 1.0
filtered_rules = rules[(rules['confidence'] >= 0.1) & (rules['lift'] > 1.0)]

print("Top Filtered Association Rules (Sorted by Lift):")
print(filtered_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
      .sort_values('lift', ascending=False).head(10), "\n")

# Task 8: Interpret rules
print("--- Sample Rule Interpretation ---")
if not filtered_rules.empty:
    sample_rule = filtered_rules.sort_values('lift', ascending=False).iloc[0]
    ant = ", ".join(list(sample_rule['antecedents']))
    con = ", ".join(list(sample_rule['consequents']))
    print(f"Rule: If a customer buys '{ant}', they are likely to buy '{con}'.")
    print(f"Support: {sample_rule['support']:.4f} (Appears in {sample_rule['support']*100:.2f}% of transactions)")
    print(f"Confidence: {sample_rule['confidence']:.4f} ({sample_rule['confidence']*100:.2f}% of customers who bought '{ant}' also bought '{con}')")
    print(f"Lift: {sample_rule['lift']:.2f} (Increases likelihood by {sample_rule['lift']:.2f} times)")
else:
    print("No rules met the filtering criteria. Try lowering the support or confidence thresholds.")

# Visualizations

# 1. Bar chart of top 20 frequent itemsets
plt.figure(figsize=(10, 6))
top_itemsets = frequent_itemsets.sort_values('support', ascending=False).head(20)
itemsets_str = top_itemsets['itemsets'].apply(lambda x: ', '.join(list(x)))
sns.barplot(x=top_itemsets['support'], y=itemsets_str, palette="viridis")
plt.title('Top 20 Frequent Itemsets by Support')
plt.xlabel('Support')
plt.ylabel('Itemset')
plt.tight_layout()
plt.show()

# 2. Support vs Confidence Plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x="support", y="confidence", size="lift", hue="lift",
                sizes=(20, 200), palette="coolwarm", data=rules)
plt.title('Support vs Confidence (Bubble Size & Color = Lift)')
plt.xlabel('Support')
plt.ylabel('Confidence')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# 3. Network graph of top 20 association rules by lift
plt.figure(figsize=(10, 10))
G = nx.DiGraph()

top_rules_network = filtered_rules.sort_values('lift', ascending=False).head(20)

for _, row in top_rules_network.iterrows():
    for antecedent in row['antecedents']:
        for consequent in row['consequents']:
            G.add_edge(antecedent, consequent, weight=row['lift'])

pos = nx.spring_layout(G, k=1.5, iterations=50)
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2500,
        edge_color='gray', width=1.5, font_size=10, font_weight='bold',
        arrows=True, arrowsize=15, alpha=0.9)

plt.title('Network Graph of Top 20 Association Rules (by Lift)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 1: Import required Python libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
print("Rohith John 24BAD100")
# Task 2: Load transactional dataset and perform preprocessing
file_path = 'Groceries_dataset.csv'
df = pd.read_csv(file_path)

# Convert transactional data into a numerical Customer-Item Matrix
# Rows = Customers, Columns = Items, Values = Purchase Counts
customer_item_matrix = pd.crosstab(df['Member_number'], df['itemDescription'])

print(f"Original High-Dimensional Data Shape: {customer_item_matrix.shape}")
print(f"(We have {customer_item_matrix.shape[0]} customers and {customer_item_matrix.shape[1]} unique items/features)\n")

# Task 3: Standardize features
# Scaling is crucial so highly frequent items (like milk) don't dominate the PCA
scaler = StandardScaler()
scaled_data = scaler.fit_transform(customer_item_matrix)

# Task 4 & 5: Apply PCA and compute principal components
# We will start by fitting PCA on all components to analyze variance
pca_full = PCA()
pca_full.fit(scaled_data)

# Task 6: Determine explained variance ratio and cumulative variance
exp_var_ratio = pca_full.explained_variance_ratio_
cum_var_ratio = np.cumsum(exp_var_ratio)

# Print the variance captured by the first 10 components
print("--- Variance Analysis (First 10 Components) ---")
for i in range(10):
    print(f"PC{i+1}: Captured {exp_var_ratio[i]*100:.2f}% | Cumulative: {cum_var_ratio[i]*100:.2f}%")

# Task 7: Reduce dimensions to 2D and 3D
pca_2d = PCA(n_components=2)
data_2d = pca_2d.fit_transform(scaled_data)

pca_3d = PCA(n_components=3)
data_3d = pca_3d.fit_transform(scaled_data)

# Create a DataFrame for the 2D data for easy plotting
df_2d = pd.DataFrame(data_2d, columns=['PC1', 'PC2'])

# Task 8: Visualize transformed data
fig = plt.figure(figsize=(15, 12))

# Visualization 1: Scree Plot (Top 50 components to keep it readable)
ax1 = fig.add_subplot(2, 2, 1)
ax1.bar(range(1, 51), exp_var_ratio[:50], alpha=0.7, color='steelblue')
ax1.set_xlabel('Principal Components')
ax1.set_ylabel('Explained Variance Ratio')
ax1.set_title('Scree Plot (Top 50 Components)')

# Visualization 2: Cumulative Variance Graph
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(range(1, len(cum_var_ratio) + 1), cum_var_ratio, color='red', linewidth=2)
# Draw a line showing how many components are needed to hit 80% variance
ax2.axhline(y=0.80, color='green', linestyle='--', label='80% Variance Threshold')
ax2.set_xlabel('Number of Components')
ax2.set_ylabel('Cumulative Explained Variance')
ax2.set_title('Cumulative Variance Graph')
ax2.legend()

# Visualization 3: 2D Scatter Plot of Principal Components
ax3 = fig.add_subplot(2, 2, 3)
sns.scatterplot(x='PC1', y='PC2', data=df_2d, ax=ax3, alpha=0.5, color='purple')
ax3.set_xlabel(f'PC1 ({exp_var_ratio[0]*100:.1f}% Variance)')
ax3.set_ylabel(f'PC2 ({exp_var_ratio[1]*100:.1f}% Variance)')
ax3.set_title('2D PCA Projection (Customer Groupings)')

# Visualization 4: 3D Scatter Plot of Principal Components
ax4 = fig.add_subplot(2, 2, 4, projection='3d')
ax4.scatter(data_3d[:, 0], data_3d[:, 1], data_3d[:, 2], alpha=0.5, color='teal')
ax4.set_xlabel('PC1')
ax4.set_ylabel('PC2')
ax4.set_zlabel('PC3')
ax4.set_title('3D PCA Projection')

plt.tight_layout()
plt.show()
```
